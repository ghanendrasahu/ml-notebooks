```python
import numpy as np   #You're importing NumPy to handle arrays and mathematical operations.
input = np.array([5, 8])  # Study hours, Sleep hours
# This array represents the features (independent variables) of one sample:
# 5 hours of study
# 8 hours of sleep
weights = np.array([0.7, 0.4])
# These are the weights assigned to each input feature. Weights control the importance of each input in the prediction.
bias = 2
# Bias allows the model to shift the activation function curve and helps with flexibility.
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
# The sigmoid function maps any value to a number between 0 and 1. It's often used in binary classification or to model probability.
weighted_sum = np.dot(input, weights) + bias
# (5 * 0.7) + (8 * 0.4) + 2 = 3.5 + 3.2 + 2 = 8.7
output = sigmoid(weighted_sum)
print(f"Weighted_Sum (z):{weighted_sum}")
print(f"Output after Sigmoid:{output}")

# This means:
# The neuron is highly activated
# The final prediction is close to 1, which is often interpreted as "yes", "true", or a high probability (like passing a test, being positive, etc.)

```

```python
import tensorflow as tf
import numpy as np

# Inputs: [study_hours, sleep_hours]
X = np.array([[5.0, 8.0]], dtype=np.float32)

# Create a Dense layer with 1 neuron, sigmoid activation
model = tf.keras.Sequential([
    tf.keras.layers.Dense(units=1, activation='sigmoid', input_shape=(2,))
])

# Manually set weights and bias (just like your example)
weights = np.array([[0.7], [0.4]], dtype=np.float32)  # Shape (2, 1)
bias = np.array([2.0], dtype=np.float32)              # Shape (1,)
model.layers[0].set_weights([weights, bias])

# Get prediction
output = model.predict(X)

print(f"Output after sigmoid activation: {output[0][0]}")
```

```python
import tensorflow as tf
import numpy as np

# Input data (1 sample with 2 features)
X = np.array([[5.0, 8.0]], dtype=np.float32)

# Build the model with 2 layers
model = tf.keras.Sequential([
    tf.keras.layers.Dense(units=2, activation='sigmoid', input_shape=(2,)),  # Hidden Layer
    tf.keras.layers.Dense(units=1, activation='sigmoid')                     # Output Layer
])

# Manually set weights for hidden layer (2 neurons)
# Each neuron has 2 weights (for 2 input features)
# Shape: (2 inputs, 2 neurons)
hidden_weights = np.array([[0.7, 0.2],
                           [0.4, 0.6]], dtype=np.float32)
hidden_bias = np.array([2.0, 1.0], dtype=np.float32)  # Shape: (2 neurons,)

# Manually set weights for output layer (from 2 hidden neurons to 1 output)
# Shape: (2 neurons, 1 output)
output_weights = np.array([[1.0],
                           [1.5]], dtype=np.float32)
output_bias = np.array([0.5], dtype=np.float32)  # Shape: (1 output neuron,)

# Set weights and biases
model.layers[0].set_weights([hidden_weights, hidden_bias])
model.layers[1].set_weights([output_weights, output_bias])

# Make prediction
output = model.predict(X)

print(f"Final Output: {output[0][0]}")
```
