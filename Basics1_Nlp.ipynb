```python
# ==========================
# TEXT PREPROCESSING IN NLP
# ==========================

# Import required libraries
import nltk
import spacy
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, SnowballStemmer

# Download NLTK resources
nltk.download('all')


# Sample text for demonstration
text = """Text preprocessing is an important step in Natural Language Processing (NLP).
It includes multiple techniques such as tokenization, stop word removal, and stemming.
These techniques help clean and prepare the text for analysis."""

```

```python

# ======================================
# 1. Tokenization (Breaking text into words/sentences)
# ======================================

print("### Tokenization ###")

# Word Tokenization using NLTK
words = word_tokenize(text)
print("\nWord Tokenization:", words)

# Sentence Tokenization using NLTK
sentences = sent_tokenize(text)
print("\nSentence Tokenization:", sentences)

# ======================================
# 2. Stop Word Removal (common words like 'is', 'the', etc.)
# ======================================

print("\n### Stop Word Removal ###")

# Load English stop words from NLTK
stop_words = set(stopwords.words('english'))

# Remove stop words from tokenized words
filtered_words = [word for word in words if word.lower() not in stop_words]
print("\nWords after Stop Word Removal:", filtered_words)

# ======================================
# 3. Stemming (Reducing words to their root form)
# ======================================

print("\n### Stemming ###")

# Initialize the Porter Stemmer
porter = PorterStemmer()
porter_stemmed = [porter.stem(word) for word in filtered_words]
print("\nPorter Stemmer Output:", porter_stemmed)

# Initialize the Snowball Stemmer (improved version)
snowball = SnowballStemmer("english")
snowball_stemmed = [snowball.stem(word) for word in filtered_words]
print("\nSnowball Stemmer Output:", snowball_stemmed)

# ======================================
# 4. spaCy for Tokenization, Stop Word Removal, and Lemmatization
# ======================================

print("\n### Using spaCy ###")

# Load the English model from spaCy
nlp = spacy.load("en_core_web_sm")

# Process the text
doc = nlp(text)

# Tokenization using spaCy (includes punctuation and special tokens)
spacy_tokens = [token.text for token in doc]
print("\nspaCy Tokenization:", spacy_tokens)

# Stop Word Removal using spaCy
spacy_filtered = [token.text for token in doc if not token.is_stop]
print("\nspaCy Stop Word Removal:", spacy_filtered)

# Lemmatization using spaCy (base form of words)
spacy_lemmatized = [token.lemma_ for token in doc if not token.is_stop]
print("\nspaCy Lemmatization:", spacy_lemmatized)

# Optional: Removing punctuation and non-alphabetic tokens
spacy_cleaned = [token.lemma_ for token in doc if not token.is_stop and token.is_alpha]
print("\nspaCy Cleaned (No Stopwords, Punctuations):", spacy_cleaned)
```
