| Function                                    | Purpose                                  |
| ------------------------------------------- | ---------------------------------------- |
| `tf.constant()`                             | Create constant tensors                  |
| `tf.Variable()`                             | Create trainable variables               |
| `tf.zeros()`, `tf.ones()`                   | Create tensors filled with zeros or ones |
| `tf.random.normal()`, `tf.random.uniform()` | Generate random tensors                  |



```python
# 1. tf.constant()
c = tf.constant([1, 2, 3])
print("Constant Tensor:", c.numpy())

# 2. tf.Variable()
v = tf.Variable([1, 2, 3])
print("Variable Tensor:", v.numpy())

# 3. tf.zeros()
z = tf.zeros([2, 3])
print("Zeros Tensor:\n", z.numpy())

# 4. tf.ones()
o = tf.ones([2, 2])
print("Ones Tensor:\n", o.numpy())

# 5. tf.random.normal()
rn = tf.random.normal([2, 2], mean=0.0, stddev=1.0)
print("Random Normal Tensor:\n", rn.numpy())

# 6. tf.random.uniform()
ru = tf.random.uniform([2, 2], minval=0, maxval=10)
print("Random Uniform Tensor:\n", ru.numpy())
```

| Function                              | Purpose                              |
| ------------------------------------- | ------------------------------------ |
| `tf.add()`, `tf.subtract()`           | Basic arithmetic                     |
| `tf.multiply()`, `tf.divide()`        | Element-wise multiplication/division |
| `tf.matmul()`                         | Matrix multiplication                |
| `tf.reduce_sum()`, `tf.reduce_mean()` | Reduction operations                 |


```python
import tensorflow as tf

# 1. tf.add()
print("Add:", tf.add(5, 3).numpy())  # Output: 8

# 2. tf.subtract()
print("Subtract:", tf.subtract(10, 4).numpy())  # Output: 6

# 3. tf.multiply()
print("Multiply:", tf.multiply([1, 2], [3, 4]).numpy())  # Output: [3, 8]

# 4. tf.divide()
print("Divide:", tf.divide([10, 20], [2, 5]).numpy())  # Output: [5.0, 4.0]

# 5. tf.matmul()
a = tf.constant([[1, 2]])
b = tf.constant([[3], [4]])
print("Matrix Multiplication:", tf.matmul(a, b).numpy())  # Output: [[11]]

# 6. tf.reduce_sum()
print("Reduce Sum:", tf.reduce_sum([1, 2, 3]).numpy())  # Output: 6

# 7. tf.reduce_mean()
print("Reduce Mean:", tf.reduce_mean([2.0, 4.0, 6.0]).numpy())  # Output: 4.0
```
