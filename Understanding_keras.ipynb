```python
 from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
# Sequential: a way to stack layers one after another.
# Dense: a fully connected layer (every neuron connects to every neuron in the previous layer).
# Create a Sequential model
model = Sequential([
    Dense(8, activation='relu', input_shape=(4,)),  # Hidden layer with 8 neurons
    Dense(1, activation='sigmoid')  # Output layer with 1 neuron
])

# Dense(8): 8 neurons in this layer
# activation='relu': ReLU is a function that helps the network learn better by introducing non-linearity
# input_shape=(4,): input has 4 features (e.g., age, income, height, weight)

# Dense(1): Only 1 neuron
# activation='sigmoid': Outputs a value between 0 and 1 — used
# for binary classification (e.g., yes/no, pass/fail, spam/not spam)

# Print model summary
model.summary()
```

How are parameters calculated?
For First Layer:
4 inputs × 8 neurons = 32 weights

8 biases = 8

→ 32 + 8 = 40 parameters

For Second Layer:
8 inputs × 1 neuron = 8 weights

1 bias = 1

→ 8 + 1 = 9 parameters

```python
# Import TensorFlow and necessary Keras modules
import tensorflow as tf
from tensorflow.keras.models import Sequential   # Sequential model is a linear stack of layers
from tensorflow.keras.layers import Dense        # Dense is a fully connected (dense) layer

# Define the model architecture using Sequential API
model = Sequential([
    # First hidden layer:
    # - 8 neurons
    # - ReLU activation function
    # - input_shape=(2,) means we have 2 input features (e.g., study hours and sleep hours)
    Dense(8, activation='relu', input_shape=(2,)),

    # Output layer:
    # - 1 neuron (binary output: yes or no)
    # - Sigmoid activation function to output probability between 0 and 1
    Dense(1, activation='sigmoid')
])

# Compile the model:
# - optimizer='adam': an advanced optimizer that adjusts learning rates automatically
# - loss='binary_crossentropy': used for binary classification (output is 0 or 1)
# - metrics=['accuracy']: track how accurate the predictions are during training
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Import NumPy for creating training data
import numpy as np

# Define the input data (X)
# Each row is a training example with 2 features: [study_hours, sleep_hours]
X = np.array([[5, 8],      # Example 1
              [2, 3]],     # Example 2
             dtype=np.float32)

# Define the target labels (y)
# Each value is the expected output (1 = pass, 0 = fail, for example)
y = np.array([[1],         # Label for Example 1
              [0]],        # Label for Example 2
             dtype=np.float32)

# Train the model
# - X: input data
# - y: true labels
# - epochs=10: repeat the training process 10 times
model.fit(X, y, epochs=10)



```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import numpy as np
# TensorFlow for deep learning
# Sequential and Dense to build the neural network
# NumPy to create and handle the dataset
np.random.seed(42)
X = np.random.rand(1000, 4)  # 1000 rows, 4 columns = 1000 samples, 4 features
# np.random.seed(42): Makes the results repeatable
# X: Random values between 0 and 1 (your input data)
y = (X[:, 0] + X[:, 1] > 1).astype(int)
# y: Target labels.
# Logic: If the sum of the first two features is greater than 1, label it as 1, else 0.
# So you're doing binary classification: 0 or 1.
model = Sequential([
    Dense(8, activation='relu', input_shape=(4,)),  # Hidden layer with 8 neurons
    Dense(1, activation='sigmoid')                  # Output layer with 1 neuron
])
# Input: 4 features
# First layer: 8 neurons, ReLU activation (good for learning complex patterns)
# Second layer: 1 neuron, sigmoid (because we're predicting probability for binary classification)
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
# Optimizer: adam — adjusts weights efficiently
# Loss: binary_crossentropy — good for binary classification
# Metric: Accuracy — how many predictions the model got right
model.fit(X, y, epochs=20, batch_size=32, verbose=1)
# epochs=20: go through the dataset 20 times
# batch_size=32: update the model after every 32 samples
# verbose=1: shows progress
```
