What is spaCy?

spaCy is an open-source NLP library designed for speed, efficiency, and real-world use.

It provides pre-trained models and tools for:

Tokenization

Part-of-speech tagging (POS)

Named Entity Recognition (NER)

Dependency parsing

Lemmatization

Sentence segmentation

Word vectors (in some models)


```python
import spacy
nlp = spacy.load("en_core_web_sm")
text = "Elon Musk founded Tesla in 2003"
doc = nlp(text)
for ent in doc.ents:
  print(ent.text,"-",ent.label_)
```

What is TextBlob?

TextBlob is a beginner-friendly Natural Language Processing (NLP) library built on top of:

NLTK (Natural Language Toolkit)

Pattern (another NLP library)

It allows you to perform common NLP tasks with just a few lines of code

```python
from textblob import TextBlob
text = "this is a bad product!"
sentiment = TextBlob(text).sentiment.polarity
if sentiment >0:
  print("Positive sentiment")
elif sentiment <0:
  print("Negative sentiment")
else:
  print("Neutral sentiment")
```

What is Counter?

Counter is a specialized dictionary subclass used to count hashable objects (like words, numbers, characters, etc.).

It's very useful for frequency analysis in tasks like text processing, NLP, or data analysis.



```python
from collections import Counter
text = "Data science is amazing. Data is the future. Science and data drive innovation"
#COnvert text to lowercase and split words
words = text.lower().split()
#Count words occurrences
word_counts = Counter(words)
print(word_counts)
```

 What is Tokenization?

Tokenization is the process of splitting text into smaller units, typically words or sentences.

```python
import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt_tab')
nltk.download('punkt')
```

```python
text = "Hello! How are you today?"
tokens = word_tokenize(text)
print(tokens)
```
