```python
import pandas as pd
left = pd.DataFrame({'key': ['A', 'B', 'C'], 'value': [1, 2, 3]})
right = pd.DataFrame({'key': ['A', 'B', 'D'], 'value': [4, 5, 6]})
print(left)
print(right)
merged = pd.merge(left, right, on='key', how='inner')
merged
```

```python
import requests

# GET request Basic Usage
response = requests.get("https://jsonplaceholder.typicode.com/posts/1")

print(response.status_code)   # HTTP status code (200 = OK)
print(response.text)          # Response as raw text
print(response.json())        # If response is JSON, convert to Python dict
```

```python
# Sending Parameters

params = {"userId": 1} # userid -1 has 10 records
response = requests.get("https://jsonplaceholder.typicode.com/posts", params=params)
print(response.json())
```

```python
#POST request
data = {"title": "Hello", "body": "World", "userId": 1}
response = requests.post("https://jsonplaceholder.typicode.com/posts", json=data)
print(response.status_code)
print(response.json())
```

```python
# Adding Headers
headers = {"Authorization": "Bearer my_token"}
response = requests.get("https://jsonplaceholder.typicode.com/posts", headers=headers)
print(response.status_code)
```

```python
# Handling Errors

try:
    response = requests.get("https://api.example.com/404")
    response.raise_for_status()  # raises exception for 4xx/5xx
except requests.exceptions.RequestException as e:
    print("Error:", e)
```

```python
data = {"title": "New Post", "body": "Hello World", "userId": 1}
response = requests.post("https://jsonplaceholder.typicode.com/posts", json=data)
print(response.status_code)
print(response.json())   # shows new post with ID
```

```python
data = {"id": 1, "title": "Updated Title", "body": "Updated Body", "userId": 1}
response = requests.put("https://jsonplaceholder.typicode.com/posts/1", json=data)
print(response.status_code)
print(response.json())
```

```python
response = requests.delete("https://jsonplaceholder.typicode.com/posts/1")
print(response.status_code)   # 200 = success, 204 = no content
```
