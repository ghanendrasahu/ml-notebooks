## Data pre-processing
Data pre-processing is a critical step in the data science workflow.

It involves preparing and transforming raw data into a suitable format that allows 
for more effective analysis and modeling.



### Why we do Data PreProcessing:

Learning algorithms have affinity towards certain pattern of data.

Unscaled or unstandardized data might have unacceptable prediction

Learning algorithms understands only number, converting text or image to number is required

Preprocessing refers to transformation of data before feeding to machine learning model

## 2. StandardScaler
* The StandardScaler assumes your data is normally distributed within each feature and will scale them such that the distribution is now centred around 0, with a standard deviation of 1.
* Calculate - Subtract mean of column & div by standard deviation



<img src="https://github.com/awantik/machine-learning-slides/blob/master/pp4.PNG?raw=true">


* If data is  normally distributed, this is not the best scaler to use.

What is Standardization?
Standardization transforms your features (columns of data) so they have:

a mean (average) of 0

a standard deviation of 1

This is useful when your features have different units/scales, and you want them to be on the same scale.

```python
# example of a standardization
from numpy import asarray
from sklearn.preprocessing import StandardScaler

# define data
data = asarray(
[[100, 0.001],
[8, 0.05],
[50, 0.005],
[88, 0.07],
[4, 0.1]])
# First column: large values (4 to 100)

# Second column: small values (0.001 to 0.1)

# define standard scaler
scaler = StandardScaler()
# transform data
scaled = scaler.fit_transform(data)
print(scaled)

# Interpretation of z-values (standardized values):
# 0 → Value is exactly the mean

# > 0 → Value is above average

# < 0 → Value is below average

#  Why use StandardScaler?
# Many ML algorithms (like Logistic Regression, SVM, KNN) perform better when features are on the same scale.

# It helps with model convergence and avoids biasing the model toward features with larger values.
```

## 3. MinMaxScaler
* One of the most popular
* Calculate - Subtract min of column & div by difference between max & min




<img src="https://github.com/awantik/machine-learning-slides/blob/master/pp3.PNG?raw=true">



* Data shifts between 0 & 1
* If distribution not suitable for StandardScaler, this scaler works out.
* Sensitive to outliers

```python

# example of a normalization
from numpy import asarray
from sklearn.preprocessing import MinMaxScaler
import pandas as pd
# define data
data = asarray([
[100, 0.001],
[8, 0.05],
[50, 0.005],
[88, 0.07],
[4, 0.1]])
# define min max scaler
scaler = MinMaxScaler()
# transform data
scaled = scaler.fit_transform(data)
print(scaled)

#  When to use MinMaxScaler?
# Use it when:

# You want to bring all features to a uniform scale

# Your data doesn’t follow a normal distribution

# You're using algorithms sensitive to feature scales (like KNN, SVM, or Neural Networks)
```

```python
df=pd.DataFrame(scaled,columns=["Marks","percentage"])
df.head()
```

#### Encoding Nominal Values
* Nominal Values - Male, Female. No relationship between data
* One Hot Encoding for converting data into one-hot vector

```python
df = pd.DataFrame({
    'Age':[33,44,22,44,55,22,35],
    'Gender':['Male','Female','Male','Female','Male','Male','LGBTQ']})

# Gender is nominal: no inherent order or ranking between the values like Male, Female, LGBTQ.

# You can't directly feed these into most ML models — they need numerical representation.
```

```python
df.Gender.unique()
```

```python
df.head(20)
```

```python
df=pd.get_dummies(df, columns=['Gender'],dtype=int)
df.head()
```
